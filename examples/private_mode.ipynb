# Training (private) mode

You get a whole environment of your own and drive all its agents (`agent_0`, `agent_1`, ...), with random actions. The token and the name are read from the `USER_TOKEN` and `USER_NAME` environment variables.

Rendering your world needs an administrator token: the picture of the world is reserved to the organizers, in the training mode as in the contest mode.

```python
# MP4 encoding needs imageio and a ffmpeg binary.
# pip install --quiet imageio imageio-ffmpeg
```

```python
import os

import imageio.v2 as imageio
import pettingzoo
from IPython.display import Video

# Importing pzclient registers "alife/alife-remote-v1" in the PettingZoo registry
import pzclient

pzclient.__version__
```

```python
env = pettingzoo.make(
    "parallel",
    "alife/alife-remote-v1",
    api_url="https://csc53439ep.jdhp.org/training/api",  # the training server
    token=os.environ["USER_TOKEN"],
    render_mode="rgb_array",
)

print(f"{os.environ['USER_NAME']} drives {env.possible_agents}")
```

```python
observations, infos = env.reset(seed=42)
frames = [env.render()]

for _ in range(100):  # the episode is eternal: bound the loop
    actions = {agent: env.action_space(agent).sample() for agent in env.agents}
    observations, rewards, terminations, truncations, infos = env.step(actions)
    frames.append(env.render())

    if not env.agents:  # only if the server sets a maximum episode length
        observations, infos = env.reset()

env.close()
```

```python
imageio.mimwrite("private_mode.mp4", frames, fps=env.metadata["render_fps"])

Video("private_mode.mp4", embed=True)
```
