# Contest (shared) mode

Your agent, named after you, plays in the world shared with the other participants, with random actions. The token and the name are read from the `USER_TOKEN` and `USER_NAME` environment variables.

Rendering the shared world needs an administrator token: in the contest mode, the picture of the world is reserved to the organizers.

```python
# MP4 encoding needs imageio and a ffmpeg binary.
# pip install --quiet imageio imageio-ffmpeg
```

```python
import os

import imageio.v2 as imageio
import pettingzoo
from IPython.display import Video

# Importing pzclient registers "alife/alife-remote-v1" in the PettingZoo registry
import pzclient

pzclient.__version__
```

```python
agent = os.environ["USER_NAME"]  # in the shared world, your agent is named after you

env = pettingzoo.make(
    "parallel",
    "alife/alife-remote-v1",
    api_url="https://csc53439ep.jdhp.org/api",
    token=os.environ["USER_TOKEN"],
    render_mode="rgb_array",
)
```

```python
observations, infos = env.reset()  # join the shared world
frames = [env.render()]

for _ in range(100):  # the episode is eternal: bound the loop
    actions = {agent: env.action_space(agent).sample()}
    observations, rewards, terminations, truncations, infos = env.step(actions)
    frames.append(env.render())

    if not env.agents:  # only if the organizers set a maximum episode length
        observations, infos = env.reset()

env.close()  # leave the shared world
```

```python
imageio.mimwrite("shared_mode.mp4", frames, fps=env.metadata["render_fps"])

Video("shared_mode.mp4", embed=True)
```
